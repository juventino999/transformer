```python
from typing_extensions import Self

import torch
import torch.nn as nn
```

```python
device = ("cuda" if torch.cuda.is_available() else "cpu")
print(device)
```

## Softmax

```python
class Softmax(nn.Module):
    def __init__(self: Self, temperature: float) -> None:
        super().__init__()
        if temperature == 0:
            raise ValueError('Temperature cannot be zero.')
        self.temperature = temperature
    def forward(self: Self, X: torch.Tensor, dim: float = 0) -> torch.Tensor:
        # add eps so there's never division by zero
        intermediate = torch.tensor(torch.e).pow(X.div(self.temperature)) + 1e-10
        # unsqueeze to align dimensions (on dim zero want something like (2, 4, 1) rather than just (2, 4))
        denominator = torch.unsqueeze(intermediate.sum(dim), dim)
        return intermediate.div(denominator)
```

```python
my_tensor = torch.tensor([
        [
            [1., 2., 0., -1e10],
            [0.04, 2, 0.5, 1.5],
        ],
        [
            [11., 2., 10., -1e10],
            [0.04, 2, 1.5, 0.5],
        ],

    ], 
    device=device
)
```

```python
dim = 0
```

```python
softmax = Softmax(1)
softmax(my_tensor, dim = dim)
```

```python
torch.nn.functional.softmax(my_tensor, dim = dim)
```
